```python
import os
import pandas as pd
import random

n = 5  # nombre de colonnes
m = 9  # nombre de valeurs par colonne

data = {'code' : [i for i in range(34000,34000 + m)]}

for i in range(n):
    column_name = f'colonne_{i+1}'
    data[column_name] = [f"mod_{i+1}{random.randint(1,m)}" for _ in range(m)]

df = pd.DataFrame(data)

for i in range(n):
    df[f'num_{i+1}'] = [random.randint(1, 100) for _ in range(m)]

file_path = os.path.join('..', 'data', 'fichier_serie.xlsx')
# df.to_excel(file_path, index=False)
df.head()
```

```python

l_object = list(df.select_dtypes('object').columns)
l_object = [col for col in l_object if col != df.columns[0]]
d = {}
for col in l_object:
    df_col = df[[df.columns[0], col]]
    df_cross = pd.crosstab(df_col[df_col.columns[0]], df_col[col])
    d[col] = df_cross

l_numeric = list(df.select_dtypes('number').columns)
l_numeric = [col for col in l_numeric if col != df.columns[0]]
d['numerique'] = df[[df.columns[0]] + l_numeric]

for col, df_cross in d.items():
    print(f"Tableau croisé pour **{col}**:")
    print(df_cross.head(2))
```

```python
file_path = os.path.join('..', 'data', 'fichier_cross.xlsx')
with pd.ExcelWriter(file_path, engine='openpyxl') as writer:
    for sheet_name, data in d.items():
        data.to_excel(writer, sheet_name=sheet_name, index=False)
```
